```python
# ============================
# 🧠 Dyna-Q implementation (CliffWalking-v0)
# ============================

# --- 環境の用意 ---
env = gym.make("CliffWalking-v0")

n_actions = env.action_space.n
n_states = env.observation_space.n

# --- ハイパーパラメータ ---
alpha = 0.1          # 学習率
gamma = 0.95         # 割引率
epsilon = 0.1        # ε-greedy探索
planning_steps = 20  # モデルベース更新回数
episodes = 200

# --- Q値とモデルを初期化 ---
Q = np.zeros((n_states, n_actions))
model = {}  # {(s, a): (r, s')}

def epsilon_greedy(state):
    if np.random.rand() < epsilon:
        return np.random.randint(n_actions)
    else:
        return np.argmax(Q[state])

rewards = []

for ep in range(episodes):
    s = np.int64(env.reset())
    total_reward = 0

    while True:
        a = epsilon_greedy(s)
        s_next, r, done, _ = env.step(a)

        # --- 実体験によるQ更新 ---
        Q[s, a] += alpha * (r + gamma * np.max(Q[s_next]) - Q[s, a])

        # --- モデル更新 ---
        model[(s, a)] = (r, s_next)

        # --- モデルからの仮想体験更新（planning） ---
        for _ in range(planning_steps):
            s_m, a_m = random.choice(list(model.keys()))
            r_m, s_next_m = model[(s_m, a_m)]
            Q[s_m, a_m] += alpha * (r_m + gamma * np.max(Q[s_next_m]) - Q[s_m, a_m])

        s = s_next
        total_reward += r
        if done:
            break

    rewards.append(total_reward)

print("✅ 学習完了！")

# --- 結果の可視化 ---
plt.figure(figsize=(8,4))
plt.plot(rewards)
plt.xlabel("Episode")
plt.ylabel("Total Reward")
plt.title("Dyna-Q on CliffWalking-v0")
plt.grid(True)
plt.show()
```
